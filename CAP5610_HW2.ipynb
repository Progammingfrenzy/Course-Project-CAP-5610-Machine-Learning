Problem 3: Computing Similarity and Distance with NumPy

```python
import numpy as np

#Two fixed vectors
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

#Cosine Similarity
cosine_similarity = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

#Euclidean distance
euclidean_distance = np.linalg.norm(A - B)


print("Cosine Similarity:", cosine_similarity)
print("Euclidean Distance:", euclidean_distance)
```

Problem 4:Linear Regression with Gradient Descent

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#Load data
data = pd.read_csv("/HW2_linear_data.csv", header=None)

X = data.iloc[:, 0].values
Y = data.iloc[:, 1].values

print(X[ :5])
print(Y[ :5])
```

```python
# Starting values
m = 0.0
c = 0.0

learning_rate = 0.0001
epochs = 1000
n = len(X)

for epoch in range(epochs):
  Y_pred = m * X + c

  dm = (2 / n) * np.sum(X * (Y_pred -Y))
  dc = (2 / n) * np.sum(Y_pred -Y)

  m = m - learning_rate * dm
  c = c - learning_rate * dc

print("Slope (m):", m)
print("Intercept(c):", c)
```

```python
Y_pred = m * X + c

mse = np.mean((Y_pred - Y) ** 2)

print("Final MSE:", mse)
```

```python
order = np.argsort(X)

plt.scatter(X, Y, label="Data")
plt.plot(X[order], Y_pred[order], label="Regression Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")
plt.legend()
plt.show()
```
