Problem 3: Computing Similarity and Distance with NumPy

```python
import numpy as np

#Two fixed vectors
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

#Cosine Similarity
cosine_similarity = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

#Euclidean distance
euclidean_distance = np.linalg.norm(A - B)


print("Cosine Similarity:", cosine_similarity)
print("Euclidean Distance:", euclidean_distance)
```

Problem 4: Linear Regression with Gradient Descent

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#Load data
data = pd.read_csv("/HW2_linear_data.csv", header=None)

X = data.iloc[:, 0].values
Y = data.iloc[:, 1].values

print(X[ :5])
print(Y[ :5])
```

```python
# Starting values
m = 0.0
c = 0.0

learning_rate = 0.0001
epochs = 1000
n = len(X)

for epoch in range(epochs):
  Y_pred = m * X + c

  dm = (2 / n) * np.sum(X * (Y_pred -Y))
  dc = (2 / n) * np.sum(Y_pred -Y)

  m = m - learning_rate * dm
  c = c - learning_rate * dc

print("Slope (m):", m)
print("Intercept(c):", c)
```

```python
Y_pred = m * X + c

mse = np.mean((Y_pred - Y) ** 2)

print("Final MSE:", mse)
```

```python
order = np.argsort(X)

plt.scatter(X, Y, label="Data")
plt.plot(X[order], Y_pred[order], label="Regression Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")
plt.legend()
plt.show()
```

Problem 5: Nonlinear Regression with Gradient Descent

```python
# Load nonlinear data
data2 = pd.read_csv("/HW2_nonlinear_data.csv")

X2 = data2["X"].values
Y2 = data2["Y"].values

print(X2[:5])
print(Y2[:5])
```

```python
# Starting values
a = 0.0
b = 0.0
c3 = 0.0
d = 0.0

learning_rate = 1e-6
epochs = 10000
n = len(X2)

for epoch in range(epochs):

    # Cubic prediction
    Y_pred2 = a * X2**3 + b * X2**2 + c3 * X2 + d

    # Error
    error = Y_pred2 - Y2

    # Gradients
    da = (2 / n) * np.sum(error * X2**3)
    db = (2 / n) * np.sum(error * X2**2)
    dc = (2 / n) * np.sum(error * X2)
    dd = (2 / n) * np.sum(error)

    # Update parameters
    a = a - learning_rate * da
    b = b - learning_rate * db
    c3 = c3 - learning_rate * dc
    d = d - learning_rate * dd

print("a:", a)
print("b:", b)
print("c:", c3)
print("d:", d)
```

```python
# Final predictions
Y_pred2 = a * X2**3 + b * X2**2 + c3 * X2 + d

# Mean Squared Error
mse2 = np.mean((Y_pred2 - Y2) ** 2)

print("Final MSE:", mse2)
```

```python
order2 = np.argsort(X2)

plt.scatter(X2, Y2, label="Data")
plt.plot(X2[order2], Y_pred2[order2], label="Cubic Regression")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Nonlinear Regression")
plt.legend()
plt.show()
```
